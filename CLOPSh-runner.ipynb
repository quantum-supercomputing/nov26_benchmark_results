```python
import time
import json
from qiskit_ibm_runtime import QiskitRuntimeService
from qiskit_device_benchmarking.clops.clops_benchmark import clops_benchmark

service = QiskitRuntimeService()
```

```python
st = time.perf_counter()
my_clops_run = clops_benchmark(service, "ibm_chicago")
ft = time.perf_counter()
classical_time = ft - st
```

```python
clops = my_clops_run.clops()
```

```python
my_clops_run.job.usage()
```

```python
results = {'name': 'clops_h', 'backend': my_clops_run.job.backend().name, 'datetime': str(my_clops_run.job.creation_date),
           'quantum_runtime': my_clops_run.job.usage(), 'classical_runtime': classical_time,
           'result': clops}
```

```python
with open("clops_h_results.json", "w") as f:
        json.dump(results, f)
```
