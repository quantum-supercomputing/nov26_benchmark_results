```python
import json
import numpy as np
import mthree
from qiskit_ibm_runtime import QiskitRuntimeService
```

```python
service = QiskitRuntimeService()
```

```python
backend = service.backend("ibm_chicago")
```

```python
mit = mthree.M3Mitigation(backend)
```

```python
jobs = mit.cals_from_system()
```

```python
fids = np.asarray(mit.readout_fidelity())
```

```python
results = {'name': 'readout', 'backend': backend.name, 'datetime': str(jobs[0].creation_date),
           'shots': 10_000, 'runtime': jobs[0].usage(),
           'readout_fids': list(float(ff) for ff in fids), 'single_qubit_cals': list(mm.tolist() for mm in mit.single_qubit_cals)}
```

```python
with open("readout_results.json", "w") as f:
        json.dump(results, f)
```
